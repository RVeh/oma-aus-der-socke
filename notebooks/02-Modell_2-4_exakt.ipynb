# Exakte Berechnungen für die Modelle 2 bis 4

In diesem Notebook werden die Wahrscheinlichkeiten $P(T=n)$ für die Modelle 2 bis 4 exakt berechnet.

Die Tabellen enthalten jeweils Brüche und zusätzlich Dezimalzahlen. Die Bruchdarstellung zeigt, dass die Werte nicht durch Simulation entstehen, sondern exakt aus der jeweiligen Modellierung berechnet werden.

Das Notebook ergänzt das Simulationsnotebook: Dort werden Daten erzeugt und ausgewertet; hier werden die theoretischen Vergleichswerte bestimmt.

```python
from fractions import Fraction
from math import comb
import pandas as pd
```

## Modell 2

In Modell 2 ergibt sich die Verteilung von $T$ aus einer negativen Binomialverteilung. Für $n\geq 3$ gilt

$$
P(T=n)=\binom{n-1}{2}\cdot \left(\frac{1}{3}\right)^3\cdot \left(\frac{2}{3}\right)^{n-3}.
$$

Die Formel beschreibt, dass der dritte passende Treffer genau im $n$-ten Zug auftritt.

```python
def modell_2_exakt(max_n):
    daten = []

    for n in range(3, max_n + 1):
        p = Fraction(comb(n - 1, 2)) * Fraction(1, 3)**3 * Fraction(2, 3)**(n - 3)

        daten.append({
            "n": n,
            "P(T=n)": p,
            "P(T=n) als Dezimalzahl": float(p)
        })

    return pd.DataFrame(daten)

print("Modell 2")
tabelle_2 = modell_2_exakt(20)
tabelle_2
```

## Modell 3

In Modell 3 lässt sich $P(T=n)$ als Summe über mögliche Zeitpunkte $i$ schreiben. Der Index $i$ beschreibt dabei den Zeitpunkt des ersten Teilereignisses.

Für $n\geq 3$ gilt

$$
P(T=n)=\sum_{i=1}^{n-2}
\left(\frac{2}{3}\right)^{i-1}\cdot \frac{1}{3}\cdot
\left(\frac{1}{2}\right)^{n-2-i}\cdot \frac{1}{2}.
$$

Diese Darstellung ist eine Faltung zweier Wartezeiten. Sie ist für den Unterricht gut lesbar, weil die einzelnen Faktoren direkt aus der Modellierung entstehen.

```python
def modell_3_exakt(max_n):
    daten = []

    for n in range(3, max_n + 1):
        p = sum(
            Fraction(2, 3)**(i - 1)
            * Fraction(1, 3)
            * Fraction(1, 2)**(n - 2 - i)
            * Fraction(1, 2)
            for i in range(1, n - 1)
        )

        daten.append({
            "n": n,
            "P(T=n)": p,
            "P(T=n) als Dezimalzahl": float(p)
        })

    return pd.DataFrame(daten)

print("Modell 3")
tabelle_3 = modell_3_exakt(20)
tabelle_3
```

## Modell 4: Strommodell

Im Strommodell wird nicht zuerst $P(T=n)$ berechnet, sondern die Überlebenswahrscheinlichkeit

$$
S(n)=P(T>n).
$$

Für die Buchstabenfolge OMA gilt wegen der fehlenden Selbstüberlappung die Rekursion

$$
S(n)=S(n-1)-\frac{1}{27}\cdot S(n-3)
$$

für $n\geq 3$ mit den Anfangswerten

$$
S(0)=S(1)=S(2)=1.
$$

Aus $S(n)$ erhält man anschließend

$$
P(T=n)=\frac{1}{27}\cdot S(n-3).
$$

Die Werte werden hier exakt als Brüche berechnet.

```python
def strom_exakt(max_n):
    # S(n) = P(T > n)
    S = [Fraction(1), Fraction(1), Fraction(1)]

    for n in range(3, max_n + 1):
        S.append(S[n-1] - Fraction(1, 27) * S[n-3])

    daten = []

    for n in range(3, max_n + 1):
        p = Fraction(1, 27) * S[n-3]

        daten.append({
            "n": n,
            "P(T=n)": p,
            "P(T=n) als Dezimalzahl": float(p),
            "P(T>n)": S[n],
            "P(T>n) als Dezimalzahl": float(S[n])
        })

    return pd.DataFrame(daten)

print("Modell 4: Strommodell")
tabelle_strom = strom_exakt(50)
tabelle_strom.head(10)
```

## Partielle Kenngrößen im Strommodell

Für das Strommodell können die Kenngrößen näherungsweise über endliche Summen berechnet werden. Für eine obere Grenze $N$ setzen wir

$$
E_N(T)=\sum_{n=3}^{N} n\cdot P(T=n)
$$

und

$$
V_N(T)=\sum_{n=3}^{N} n^2\cdot P(T=n)-E_N(T)^2.
$$

Diese Größen sind partielle Näherungen. Sie verwenden nur die Werte bis $N$. Die Restwahrscheinlichkeit

$$
P(T>N)=S(N)
$$

zeigt, wie viel Wahrscheinlichkeitsmasse noch nicht berücksichtigt wurde.

Wichtig: $V_N(T)$ ist hier nicht die Varianz einer auf $\{3,\dots,N\}$ normierten, abgeschnittenen Verteilung, sondern eine partielle Näherung an die Varianz von $T$.

```python
def strom_momente_partiell(max_n):
    tabelle = strom_exakt(max_n)

    E = sum(Fraction(n) * p for n, p in zip(tabelle["n"], tabelle["P(T=n)"]))
    E2 = sum(Fraction(n*n) * p for n, p in zip(tabelle["n"], tabelle["P(T=n)"]))

    V = E2 - E**2
    rest = tabelle.iloc[-1]["P(T>n)"]

    return E, V, rest

N = 200
E_N, V_N, rest_N = strom_momente_partiell(N)

print(f"E_{N}(T) = {E_N}")
print(f"E_{N}(T) = {float(E_N):.5f} als Dezimalzahl (gerundet)")
print()
print(f"V_{N}(T) = {V_N}")
print(f"V_{N}(T) = {float(V_N):.5f} als Dezimalzahl (gerundet)")
print()
print(f"P(T>{N}) = {rest_N}")
print(f"P(T>{N}) = {float(rest_N):.8f} als Dezimalzahl (gerundet)")
```

Die exakten Tabellenwerte können im Simulationsnotebook als Vergleichswerte verwendet werden. Die Simulation liefert relative Häufigkeiten $\hat p_n$, während dieses Notebook die theoretischen Werte $P(T=n)$ berechnet.

Damit lassen sich beide Blickrichtungen vergleichen:

$$
\text{Simulation: Daten } \longrightarrow \text{ Schätzung}
$$

und

$$
\text{Modellierung: Rekursion oder Formel } \longrightarrow \text{ exakte Werte}.
